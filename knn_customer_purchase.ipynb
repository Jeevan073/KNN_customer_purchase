**PROBLEM - 1**

```python
import pandas as pd
import numpy as np
data = {
    "Distance_km":[2,3,4,5,6,7,8,9,11,13,15,18,20],
    "Duration_min":[8,12,15,20,30,35,40,50,60,90,120,125,135],
    "Category":["Short","Short","Short","Short","Medium","Medium","Medium","Long","Long","Long","Long","Long","Long"]
}
df = pd.DataFrame(data)
print(df)
```

```python
X = df[["Distance_km", "Duration_min"]]
y = df["Category"]
```

```python
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X,y)
```

```python
distance = float(input("Enter distance in km: "))
duration = float(input("Enter duration in minutes: "))

prediction = model.predict([[distance, duration]])
print("Predicted category:", prediction[0])
```

**PROBLEM - 2**

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier

data = {
    "Age": [20, 22, 25, 28, 30, 32, 35, 40],
    "Previous_Purchases": [1, 2, 3, 5, 6, 7, 8, 10],
    "Purchased": ["No", "No", "No", "Yes", "Yes", "Yes", "Yes", "Yes"]
}

df = pd.DataFrame(data)

print(df)
```

```python
X = df[["Age", "Previous_Purchases"]]
y = df["Purchased"]

model = KNeighborsClassifier(n_neighbors = 3)
model.fit(X,y)
```

```python
new_cust = [[27,4]]
prediction = model.predict(new_cust)
print("Prediction:", prediction[0])
```

```python
for k in [1,3,5]:
  model = KNeighborsClassifier(n_neighbors = k)
  model.fit(X,y)
  predictions = model.predict([[24,8]])
  print("Prediction for K =",k, "-->",predictions[0])
```
